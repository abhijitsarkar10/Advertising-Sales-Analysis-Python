```python
# load data
df1=pd.read_csv("/content/drive/MyDrive/DBA/MODULE 2/RegressionData/fev.txt",sep='\t')
pd.options.display.max_columns=10
df1.head()
```

```python
# Reload original df1 to ensure 'Sex' and 'Smoker' columns are present
df1=pd.read_csv("/content/drive/MyDrive/DBA/MODULE 2/RegressionData/fev.txt",sep='\t')
# change the categorical variables into dummy
df1=pd.get_dummies(df1,columns=['Sex','Smoker'])
df1.head()
```

```python

# Split the data into train and test
train=df1.sample(frac=0.7,random_state=0)
test=df1.drop(train.index,axis=0)
```

```python

# Split the data into train and test
train=df1.sample(frac=0.7,random_state=0)
test=df1.drop(train.index,axis=0)
```

```python
# Build a multiple regression model
reg3 = smf.ols('FEV ~ Age + Height +Sex_Male+Smoker_Non', data=train).fit()

# inspect the regression results
print(reg3.summary())
```

```python
# Check model assumptions

# homocedasticity and linearity check
plt.scatter(reg3.fittedvalues,reg3.resid)
plt.xlabel("Predicted y values")
plt.ylabel("Residuals")
```

```python
# normality check
stats.shapiro(reg3.resid) # Ho: Data is normally distributed
```

```python
# log transformation of dependent variable
# to correct homoscedasticity violation
reg4=smf.ols('np.log(FEV) ~ Age + Height +Sex_Male+Smoker_Non', data=train).fit()
print(reg4.summary())
```

```python
# check the assumptions
# independence of errors -> Durbin watson statistics
# homocedasticity and linearity check
plt.scatter(reg4.fittedvalues,reg4.resid)
plt.xlabel("Predicted y values")
plt.ylabel("Residuals")
```

```python
# normality check
stats.shapiro(reg4.resid)
```

```python
# compare performance(RMSE) of two models
X_test=test.iloc[:,[1,3,5,7]]
y_pred3=reg3.predict(X_test)
y_pred4=np.exp(reg4.predict(X_test)) # convert to original unit
```

```python
mean_squared_error(test.FEV,y_pred3)**0.5  # RMSE of model3
```

```python
mean_squared_error(test.FEV,y_pred4)**0.5  # RMSE of model4
```

```python
#%%
# Multiple regression  (with  multicollinearity problem)
import pandas as pd
import numpy as np
import statsmodels.api as sm
from statsmodels.stats.outliers_influence import variance_inflation_factor
```

```python

# load data
df=pd.read_csv("/content/drive/MyDrive/DBA/MODULE 2/RegressionData/Concrete.csv")
pd.options.display.max_columns=15 # see all columns
df.head(2)
```

```python
# rename some columns
df=df.rename(columns={'Cement ':'Cement',
                   'Slag ':'Slag',
                   'FlyAsh ':'FlyAsh',
                   'Water ':'Water',
                   'Superplasticizer ':'Superplasticizer',
                   'CoarseAggregate  ':'CoarseAggregate',
                   'FineAggregate ':'FineAggregate',
                   'Age ':'Age'})
df.head(2)
```

```python
# target and features
y=df['ConcreteStrength']
X=df.iloc[:,:-1]
X = sm.add_constant(X, prepend=False)

# Fit a regression model
res = sm.OLS(y, X).fit()
print(res.summary())
```

```python
# condition Number>3 -> strong Multicollinearity


# check for multicollinearity
vif=[variance_inflation_factor(X.values, i) for i in range(X.shape[1])]
pd.Series(vif,index=X.columns)
```

```python
# delete the features with high vif value (cutoff=5) one by one ( not all at a time)
# to  get a parsimonius and interpretable model
del X['Slag']

# recalculate vif after deleting feature with high vif
vif=[variance_inflation_factor(X.values, i) for i in range(X.shape[1])]
pd.Series(vif,index=X.columns)
```

```python
# Fit a new regression model
res1 = sm.OLS(y, X).fit()
print(res1.summary())
```
